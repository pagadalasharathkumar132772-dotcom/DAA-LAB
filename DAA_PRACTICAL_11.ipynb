 Floyd-Warshal algorithm.


```python

INF = 999

graph = [
    [0,   2,   3],
    [INF, 0,   1],
    [INF, INF, 0]
]

n = len(graph)
dist = [row[:] for row in graph]

for k in range(n):
    for i in range(n):
        for j in range(n):
            dist[i][j] = min(
                dist[i][j],
                dist[i][k] + dist[k][j]
            )

print("Shortest Distance Matrix:")
for row in dist:
    print(row)
```
