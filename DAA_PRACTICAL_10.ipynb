Implement Kruskal’s algorithm.

```python
graph = [
    ['A', 'B', 2],
    ['A', 'D', 1],
    ['D', 'C', 3],
    ['B', 'C', 6],
]

v = ['A', 'B', 'C', 'D']

parent = {x: x for x in v}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    parent[find(a)] = find(b)

graph.sort(key=lambda x: x[2])

mst = []
total = 0

for a, b, w in graph:
    if find(a) != find(b):
        union(a, b)
        mst.append((a, b, w))
        total += w

print("MST Edges:")

for a, b, w in mst:
    print(a, "-", b, ":", w)

print("Minimum Cost:", total)
```
