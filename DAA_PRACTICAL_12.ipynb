Travelling Salesman Problem

```python

from itertools import permutations

graph = [
    [0,  20, 40, 30],
    [20, 0,  10, 60],
    [40, 10, 0,  50],
    [30, 60, 50, 0]
]

cities = [1, 2, 3]
min_cost = float('inf')
best_path = []

for p in permutations(cities):
    path = [0] + list(p) + [0]
    cost = 0

    for i in range(len(path) - 1):
        cost += graph[path[i]][path[i + 1]]

    if cost < min_cost:
        min_cost = cost
        best_path = path

names = ['A', 'B', 'C', 'D']
print("Best Tour:", " -> ".join(names[i] for i in best_path))
print("Minimum Cost:", min_cost)
```
